# S3K assessment transfer
Run the cell once to copy the verified PDF and narrated video to Ravi Kumar Rao's Google Drive.

```python
from google.colab import drive
from pathlib import Path
import hashlib
import requests

drive.mount('/content/drive')
folder = Path('/content/drive/MyDrive/S3K_Internship_2026_Ravi_Kumar_Rao')
folder.mkdir(parents=True, exist_ok=True)
base = 'https://pharmacies-swaziland-stored-released.trycloudflare.com/'
files = {
    'S3K_Ravi_Kumar_Rao_Case_Study_01.pdf': ('7cf74eefb07a00f6f70649b03cbaa6b57e30d01eb3cf223ec3f509319fcce4a2', 258222),
    'MYAIGURU_case_study_Ravi_voice.mp4': ('46b6809fa5af15da7d764707d6dc6732a0712b4d00004ca2d0da255a39d0610d', 13775703),
}
for name, (expected_hash, expected_size) in files.items():
    response = requests.get(base + name, timeout=120)
    response.raise_for_status()
    data = response.content
    assert len(data) == expected_size, f'{name}: unexpected size'
    assert hashlib.sha256(data).hexdigest() == expected_hash, f'{name}: hash mismatch'
    target = folder / name
    target.write_bytes(data)
    print(f'Uploaded: {target} ({len(data)} bytes)')
```
